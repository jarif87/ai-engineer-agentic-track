```python
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from groq import Groq
from IPython.display import Markdown, display
from kaggle_secrets import UserSecretsClient
from google import genai
import os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"  # suppress TensorFlow logs if any
os.environ["CUDA_LAUNCH_BLOCKING"] = "1"  # optional: more consistent CUDA debugging

import warnings
warnings.filterwarnings("ignore", category=UserWarning)  # ignore user warnings
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)


user_secrets = UserSecretsClient()
```

```python
gemini_api_key = user_secrets.get_secret("GEMINI_API_KEY")
groq_api_key = user_secrets.get_secret("GROQ_API_KEY")

groq_client = Groq(api_key=groq_api_key)
gemini_client=genai.Client(api_key=gemini_api_key)
```

```python
request = (
    "Please come up with a challenging, nuanced question that I can ask a number "
    "of LLMs to evaluate their intelligence. Answer only with the question, no explanation."
)

messages = [{"role": "user", "content": request}]


response = groq_client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=messages,
    temperature=1,
    max_completion_tokens=1024,
    top_p=1,
    stream=True,
    stop=None
)

question = ""
for chunk in response:
    question += chunk.choices[0].delta.content or ""
    print(chunk.choices[0].delta.content or "", end="")  
    
print(f"Question is: {question}")
```

```python
competitors = []
answers = []
messages = [{"role": "user", "content": question}]
```

```python
competitors = []
answers = []

messages = [{"role": "user", "content": question}]

models_to_test = [
    "llama-3.3-70b-versatile",
    "meta-llama/llama-guard-4-12b",
    "qwen/qwen3-32b"
]

for model_name in models_to_test:
    response = groq_client.chat.completions.create(
        model=model_name,
        messages=messages,
        temperature=1,
        max_completion_tokens=1024,
        top_p=1,
        stream=False
    )
    answer = response.choices[0].message.content
    display(Markdown(f"**Model:** {model_name}\n\n{answer}"))
    
    competitors.append(model_name)
    answers.append(answer)
```

```python
# Generate Gemini response
response = gemini_client.models.generate_content(
    model="gemini-2.5-flash",
    contents=question
)

gemini_answer = response.text
display(Markdown(f"**Model:** gemini-2.5-flash\n\n{gemini_answer}"))

# Append Gemini to your lists
competitors.append("gemini-2.5-flash")
answers.append(gemini_answer)
```

```python
from sentence_transformers import SentenceTransformer
import torch

warnings.filterwarnings("ignore")
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
os.environ["CUDA_VISIBLE_DEVICES"] = "0"
os.environ["NCCL_DEBUG"] = "WARN"
os.environ["CUBLAS_WORKSPACE_CONFIG"] = ":4096:8"
os.environ["CUDA_DEVICE_ORDER"] = "PCI_BUS_ID"
os.environ["ABSL_MIN_LOG_LEVEL"] = "3"
warnings.simplefilter("ignore")


# Load embedding model
embed_model = SentenceTransformer("all-MiniLM-L6-v2")

# Generate embeddings
embeddings = embed_model.encode(answers, convert_to_tensor=True)


def cosine_sim(a, b):
    return torch.nn.functional.cosine_similarity(a, b, dim=0).item()

# Compare all pairs
for i in range(len(answers)):
    for j in range(i+1, len(answers)):
        sim = cosine_sim(embeddings[i], embeddings[j])
        print(f"{competitors[i]} vs {competitors[j]}: {sim:.3f}")
```

```python
comparison_prompt = f"""
You are an expert AI evaluator. Here are four model answers to the same question:

1. {competitors[0]}: {answers[0]}
2. {competitors[1]}: {answers[1]}
3. {competitors[2]}: {answers[2]}
4. {competitors[3]}: {answers[3]}

Rank them from best to worst based on clarity, creativity, and correctness. Provide a short explanation for your ranking.
"""

gemini_response = gemini_client.models.generate_content(
    model="gemini-2.5-flash",
    contents=comparison_prompt
)

display(Markdown(gemini_response.text))
```
