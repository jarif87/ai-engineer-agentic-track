```python
from dotenv import load_dotenv
from agents import Agent, Runner, trace
from agents.mcp import MCPServerStdio
import os
load_dotenv(override=True)
```

```python
playwright_params = {"command": "npx", "args": ["@playwright/mcp@latest"]}

mcp_browser = MCPServerStdio(params=playwright_params, client_session_timeout_seconds=600)
await mcp_browser.connect()

instructions = """
You are an elite autonomous web researcher. 
You independently browse, navigate, accept cookies, close popups, scroll, and extract content from any website.
You never give up. You try multiple sources until the result is perfect.
You output clean, structured markdown when asked.
"""
```

```python
agent = Agent(
    name="WebScout-X1",
    instructions=instructions,
    model="gpt-4o-mini",
    mcp_servers=[mcp_browser],
    tools=[]
)
```

```python
with trace("ultimate-dessert-guide"):
    result = await Runner.run(agent, """
    .output_format: markdown_only

    Mission: Create the definitive "2025 Top 5 No-Bake Summer Desserts" guide.

    For each dessert:
    1. Search at least 3 high-traffic recipe sites (BBC, Allrecipes, Sally's, Delish, etc.)
    2. Find the highest-rated version
    3. Extract and merge the best elements
    4. Include: name, origin, rating, prep time, servings, difficulty

    Desserts must be:
    - Truly no-bake (no oven, no stovetop toffee)
    - Ready under 4 hours
    - Under 400 kcal per serving

    Final output structure:
    # 2025 Top 5 No-Bake Summer Desserts
    ## 1. [Dessert Name] ★9.7
    ### Why it's #1
    ### Ingredients
    ### Instructions
    ---
    (repeat for 2–5)
    """)
    print(result.final_output.strip())
```

```python
with trace("shopping-list"):
    result = await Runner.run(agent, f"""
    .output_format: markdown_only

    You just created the "2025 Top 5 No-Bake Summer Desserts" guide.
    Now generate a combined, optimized shopping list for making ALL 5 desserts at once (for 8 people each).

    Rules:
    - Merge duplicate ingredients
    - Convert everything to grams/ml where possible
    - Group by supermarket aisle/section
    - Add checkboxes
    - Estimate total cost in USD and GBP (use current average prices)

    Title: # Master Shopping List – All 5 Desserts (40 servings)
    """)
    print(result.final_output.strip())
```

```python
with trace("final-booklet"):
    result = await Runner.run(agent, """
    .output_format: markdown_only

    Combine everything into a beautiful, printable dessert booklet.

    Title: "Summer 2025: The No-Bake Dessert Bible"
    Date: November 2025

    Must include:
    - Elegant cover page with title + date
    - Table of contents with # anchors
    - One full recipe per section with photo placeholder
    - All 5 desserts fully detailed
    - Final combined shopping list
    - Use emojis, horizontal rules ---, bold, italics, code blocks where needed
    """)

    final_md = result.final_output.strip()

    # AUTO-SAVE
    filename = "Summer_2025_NoBake_Dessert_Bible.md"
    with open(filename, "w", encoding="utf-8") as f:
        f.write(final_md)

    print(f"SAVED → {filename}")
    print("\n" + "="*60 + "\n")
    print(final_md)
```
